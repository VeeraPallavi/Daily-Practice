```python
"""Scenario 1: Daily Temperature Monitoring
You are working for a weather analytics company.
Tasks:
Create temperature data for 30 days using NumPy (random but realistic).
Store it in a Pandas DataFrame with dates.
Find:
Average temperature
Hottest and coldest day
Plot:
Line chart of temperature vs date
Highlight hottest and coldest points
"""
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Create temperature data for 30 days using NumPy (random but realistic).
#Store it in a Pandas DataFrame with dates.
np.random.seed(42)
temperature = np.random.randint(-20, 50, size = 30)
temperature
```

```python
dates = pd.date_range('1-1-2026', periods = 30) 
dates
```

```python
data = {
    'date' : pd.date_range('1-1-2026', periods = 30),
    'temperature' : np.random.randint(-20, 50, size = 30),
}

df = pd.DataFrame(data)
df
```

```python
#Average temperature
average_temperature = df['temperature'].mean()
average_temperature
```

```python
#Hottest and coldest day
hottest_day = df['temperature'].max()
print(f"Hottest Temparture : {hottest_day}")
print(f"Hottest Day : {np.argmax(df['temperature'])}")
coolest_day = df['temperature'].min()
print(f"Coolest Temperature : {coolest_day}")
print(f"Coolest Day : {np.argmin(df['temperature'])}")
```

```python
#Line chart of temperature vs date
plt.plot(df['date'],df['temperature'], color = 'red')
plt.xlabel('Date')
plt.ylabel('Temperature')
plt.title("Temperature vs Date")
plt.show()
```

```python
#Highlight hottest and coldest points
plt.boxplot(df['temperature'])
plt.xlabel("Dates")
plt.ylabel("Temperature")
plt.title("Boxplot of Temperature vs dates")
plt.show()
```
